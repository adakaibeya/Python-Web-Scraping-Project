```python
import pandas as  pd
import numpy as np
import matplotlib.pyplot as plt
from bs4 import BeautifulSoup
import requests
```

```python
url = 'https://en.wikipedia.org/wiki/List_of_largest_companies_in_Nigeria'
headers = {'User-Agent':'MyDataResearchProject/1.0(student research)'}
page = requests.get(url, headers=headers)
soup = BeautifulSoup(page.text, 'html.parser')
print(soup)
```

```python
table= soup.find_all('table')[0]
table
table_title = table.find_all('th')
table_title
```

```python
table_title_headers = [title.text.strip() for title in table_title]
table_title_headers
df= pd.DataFrame(columns=table_title_headers)
df
```
